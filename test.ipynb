```python
import numpy as np
from pytket.circuit import Circuit as PytketCircuit
from qiskit import QuantumCircuit as QiskitCircuit

from pytket.circuit.display import render_circuit_jupyter
```

```python
def get_QFT_circuit_qiskit(num_qubits):
    circ = QiskitCircuit(num_qubits, name="$$QFT$$")
    for i in range(num_qubits):
        circ.h(num_qubits-1-i)
        for j in range(i + 1, num_qubits):
            circ.cp(np.pi / 2 ** (j - i), num_qubits-1-i, num_qubits-1-j)
    return circ

get_QFT_circuit_qiskit(4).draw(fold=100)
```

```python
def get_QFT_circuit_pytket(num_qubits):
    circ = PytketCircuit(num_qubits, name="$$QFT$$")
    for i in range(num_qubits):
        circ.H(num_qubits-1-i)
        for j in range(i + 1, num_qubits):
            circ.CU1(1 / 2 ** (j - i), num_qubits-1-i, num_qubits-1-j)
    return circ

render_circuit_jupyter(get_QFT_circuit_pytket(4))
```

```python
circ = PytketCircuit(2)
circ.CU1(1/8,0,1)
u = circ.get_unitary()

for i in range(4):
    for j in range(4):
        print(u[i,j], end="\t")
    print()
```

```python
circ = QiskitCircuit(2)
circ.cp(np.pi/8,0,1)

import qiskit.quantum_info as qi

op = qi.Operator(circ)
op.data


for i in range(4):
    for j in range(4):
        print(op.data[i,j], end="\t")
    print()
```
